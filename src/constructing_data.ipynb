# Classes

```python
possible_classes = ["cat", "dog", "duck", "horse", "chicken", "car", "truck", "auto", "bike", "motorcycle"]
```

```python
annotations
```

```python
class_to_uids = {}

for classe in possible_classes:
    classe_lower = classe.lower()
    
    cc_by_uids = []
    for uid, annotation in annotations.items():

        tags_list = annotation.get("tags", [])
        tag_matches = False
        
        for tag in tags_list:

            if isinstance(tag, dict):

                tag_str = tag.get("name", "") or tag.get("value", "") or str(tag)
            else:
                tag_str = str(tag)
                
            if classe_lower == tag_str.lower():
                tag_matches = True
                break
        
        name_str = annotation.get("name", "")
        if not isinstance(name_str, str):
            name_str = str(name_str)
            
        if tag_matches or (classe_lower in name_str.lower()):
            cc_by_uids.append(uid)
            
    class_to_uids[classe] = cc_by_uids

print({k: len(v) for k, v in class_to_uids.items()})
num = []
for k, v in class_to_uids.items():
    num.append(len(v)) 

print(sum(num))
```

```python
import trimesh

mesh = trimesh.load(list(objects.values())[95], force='mesh')


pontos, face_indices = trimesh.sample.sample_surface_even(mesh, count=1024)

if len(pontos) < 1024:
    pontos, face_indices = trimesh.sample.sample_surface(mesh, count=1024)

normais = mesh.face_normals[face_indices]
```

```python
print(f"Nuvem de pontos gerada! Formato da matriz (Shape): {pontos.shape}")

normais = mesh.face_normals[face_indices]
print(f"Formato das normais (Shape): {normais.shape}")


for i in range(5):
    print(f"Ponto {i}: X={pontos[i][0]:.4f}, Y={pontos[i][1]:.4f}, Z={pontos[i][2]:.4f}")


nuvem_de_pontos = trimesh.points.PointCloud(pontos)


nuvem_de_pontos.show()
```

```python
import open3d as o3d

print("3. Criando a Triangulação de Delaunay (Alpha Shape)...")

pcd = o3d.geometry.PointCloud()
pcd.points = o3d.utility.Vector3dVector(pontos)
pcd.normals = o3d.utility.Vector3dVector(normais)
pcd.paint_uniform_color([0.2, 0.5, 0.8])


distancias = pcd.compute_nearest_neighbor_distance()
distancia_media = np.mean(distancias)


alpha = distancia_media*3
print(f"Distância média entre pontos: {distancia_media:.4f} -> Alpha definido automaticamente para: {alpha:.4f}")


mesh_delaunay = o3d.geometry.TriangleMesh.create_from_point_cloud_alpha_shape(pcd, alpha)
mesh_delaunay = o3d.geometry.TriangleMesh.create_from_point_cloud_alpha_shape(pcd, alpha)


mesh_delaunay.compute_vertex_normals()

print(f"Malha gerada com sucesso via Delaunay! Triângulos: {len(mesh_delaunay.triangles)}")


o3d.visualization.draw_geometries(
    [mesh_delaunay], 
    window_name="Malha de Delaunay (Alpha Shape) - Open3D",
    mesh_show_wireframe=True 
)
```
